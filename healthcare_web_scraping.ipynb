```python
# Open and read the downloaded XML file
with open("healthcare_data.xml", "rb") as file:
    xml_content = file.read()
# Parse the XML content
soup = BeautifulSoup(xml_content, "xml")
print("XML file loaded successfully!")
```

```python
# Get the first health-topic record
first_topic = soup.find("health-topic")
print(first_topic.prettify()[:3000])
```

```python
# Display all child element names
for element in first_topic.find_all(recursive=False):
    print(element.name)
```

```python
# Inspect selected XML elements
elements_to_inspect = [
    "also-called",
    "group",
    "language-mapped-topic",
    "mesh-heading",
    "other-language",
    "primary-institute",
    "see-reference",
    "site"
]

for element_name in elements_to_inspect:
    element = first_topic.find(element_name)
    if element:
        print(f"\n--- {element_name.upper()} ---")
        print("Attributes:", element.attrs)
        print("Text:", element.get_text(" ", strip=True)[:300])
```

```python
elements_to_inspect = [
    "also-called",
    "group",
    "mesh-heading",
    "see-reference",
    "site"
]

# Find topics with multiple elements
for element_name in elements_to_inspect:
    elements = first_topic.find_all(element_name)
    print(f"\n--- {element_name.upper()} ---")
    print("Number of elements:", len(elements))
    for element in elements:
        print("Attributes:", element.attrs)
        print("Text", element.get_text(" ", strip=True))
```

```python
from collections import Counter

# Get all health-topic records
all_topics = soup.find_all("health-topic")
print("Total health-topic records:", len(all_topics))

# Count records by language
language_counts = Counter(topic.get("language") for topic in all_topics)

print("\nRecords by language:")
for language, count in language_counts.items():
    print(language, ":", count)

# Check how many records contain each element
elements_to_check = [
    "also-called",
    "group",
    "full-summary",
    "mesh-heading",
    "language-mapped-topic",
    "other-language",
    "primary-institute",
    "see-reference",
    "site",
    "related-topic"
]

print("\nELEMENT AVAILABILITY")

for element_name in elements_to_check:
    count = sum(
        1 for topic in all_topics
        if topic.find(element_name))
    print(f"{element_name}: {count} records")
```

```python
# Filter English health-topic records
english_topics = [
    topic for topic in all_topics
    if topic.get("language") == "English"]
print("English health-topic records:", len(english_topics))

# Find an English record containing related-topic
topic_with_related = next(
    (
        topic for topic in english_topics
        if topic.find("related-topic")
    ),
    None)

# Inspect the related-topic structure
if topic_with_related:
    print("\nCondition:", topic_with_related.get("title"))
    print("\nRelated-topic elements:")
    print(topic_with_related.find("related-topic").prettify())
else:
    print("No related-topic element found.")
```

```python
# Count English records containing related topics
english_related_count = sum(
    1 for topic in english_topics
    if topic.find("related-topic"))

# Count all individual related-topic elements
total_related_elements = sum(
    len(topic.find_all("related-topic"))
    for topic in english_topics)

print("English records with related topics:", english_related_count)
print("Total related-topic associations:", total_related_elements)
```

```python
# Find an English health topic containing site elements
topic_with_sites = next(
    (topic for topic in english_topics
        if topic.find("site")),
    None)

# Inspecting the first site element
if topic_with_sites:
    first_site = topic_with_sites.find("site")

    print("Condition:", topic_with_sites.get("title"))
    print("\nSite element structure:")
    print(first_site.prettify())
```

```python
# Count all site resources in English records
total_sites = sum(
    len(topic.find_all("site"))
    for topic in english_topics)
print("Total healthcare resources:", total_sites)
```

```python
# Find all unique child element names within site elements
site_child_elements = set()

# Find all unique attributes used by site elements
site_attributes = set()

for topic in english_topics:
    for site in topic.find_all("site"):
        for child in site.find_all(recursive=False):
            site_child_elements.add(child.name)
        site_attributes.update(site.attrs.keys())

print("Site child elements:")
print(sorted(site_child_elements))

print("\nSite attributes:")
print(sorted(site_attributes))
```

```python
# Elements we want to check
child_elements = [
    "information-category",
    "organization",
    "standard-description"
]

for element_name in child_elements:
    highest_count = 0
    for topic in english_topics:
        # Find all site resources for this condition
        sites = topic.find_all("site")
        for site in sites:
            # Count how many times this element appears in the site
            count = len(site.find_all(element_name, recursive=False))
            if count > highest_count:
                highest_count = count
    print(element_name, ":", "maximum of", highest_count, "per resource")
```

```python
import pandas as pd

health_conditions = []

for topic in english_topics:
    # Extracting the condition details
    condition_id = topic.get("id")
    condition_name = topic.get("title")
    date_created = topic.get("date-created")
    short_description = topic.get("meta-desc")

    full_summary_element = topic.find("full-summary")
    if full_summary_element:
        full_summary = BeautifulSoup(full_summary_element.get_text(),"html.parser").get_text(" ", strip=True)
        full_summary = re.sub(r"\s+", " ", full_summary)
    else:
        full_summary = None

    # Extract alternative names
    alternative_names = []
    also_called_elements = topic.find_all("also-called")

    for element in also_called_elements:
        alternative_names.append(element.get_text(strip=True))
    alternative_names = "; ".join(alternative_names)

 # Extract health categories
    health_categories = []
    group_elements = topic.find_all("group")

    for element in group_elements:
        health_categories.append(element.get_text(strip=True))
    health_categories = "; ".join(health_categories)

        # Extract mesh terms
    mesh_terms = []
    mesh_heading = topic.find("mesh-heading")

    if mesh_heading:
        descriptors = mesh_heading.find_all("descriptor")

        for descriptor in descriptors:
            term = descriptor.get_text(strip=True)
            mesh_terms.append(term)
    mesh_terms = "; ".join(mesh_terms)

    mesh_ids = []
    if mesh_heading:
        for descriptor in descriptors:
            mesh_id = descriptor.get("id")
            if mesh_id:
                mesh_ids.append(mesh_id)

    mesh_ids = "; ".join(mesh_ids)
    condition_url = topic.get("url")

    condition = {
    "Condition_ID": condition_id,
    "Condition_Name": condition_name,
    "Date_Created": date_created,
    "Short_Description": short_description,
    "Full_Summary": full_summary,
    "Alternative_Names": alternative_names,
    "Health_Category": health_categories,
    "MeSH_Terms": mesh_terms,
    "MeSH_IDs": mesh_ids,
    "Condition_URL": condition_url
}
    
    health_conditions.append(condition)

health_conditions_df = pd.DataFrame(health_conditions)
health_conditions_df.head()
```

```python
print(
    "Duplicate Condition IDs:",
    health_conditions_df["Condition_ID"].duplicated().sum()
)
```

```python
# check for empty values
text_columns = [
    "Condition_Name",
    "Short_Description",
    "Full_Summary",
    "Condition_URL"]

for column in text_columns:
    empty_count = (
        health_conditions_df[column]
        .fillna("")
        .str.strip()
        .eq("")
        .sum())
    print(column, ":", empty_count, "empty values")
```

```python
# check for missing dates
health_conditions_df["Date_Created"] = pd.to_datetime(
    health_conditions_df["Date_Created"],
    format="%m/%d/%Y",
    errors="coerce"
)
print("Missing or invalid dates:",
      health_conditions_df["Date_Created"].isnull().sum())
```

```python
# Find records with an empty short description
missing_description = health_conditions_df[
    health_conditions_df["Short_Description"].isnull() |
    (health_conditions_df["Short_Description"].str.strip() == "")
]
display(missing_description)
```

```python
related_topics = []
for topic in english_topics:
    related_elements = topic.find_all("related-topic")
    for related in related_elements:

        condition_id = topic.get("id")
        condition_name = topic.get("title")
        related_topic_id = related.get("id")
        related_topic_name = related.get_text(strip=True)
        related_topic_url = related.get("url")

        record = {
            "Condition_ID": condition_id,
            "Condition_Name": condition_name,
            "Related_Topic_ID": related_topic_id,
            "Related_Topic_Name": related_topic_name,
            "Related_Topic_URL": related_topic_url}

        related_topics.append(record)
        
related_topics_df = pd.DataFrame(related_topics)
display(related_topics_df.head())
print("Total related-topic records:", len(related_topics_df))
print("Total columns:", len(related_topics_df.columns))
```

```python
related_topics_df.insert(
    0,
    "Related_ID",
    range(1, len(related_topics_df) + 1)
)
related_topics_df["Main_Topic_Language"] = "English"
related_topics_df["Source_Name"] = "MedlinePlus"
display(related_topics_df.head())
```

```python
healthcare_resources = []
for topic in english_topics:
    site_elements = topic.find_all("site")
    
    for site in site_elements:
        condition_id = topic.get("id")
        condition_name = topic.get("title")
        resource_title = site.get("title")
        resource_url = site.get("url")
        language_mapped_url = site.get("language-mapped-url")

        information_categories = []
        for element in site.find_all("information-category", recursive=False):
            information_categories.append(element.get_text(strip=True))

        organizations = []
        for element in site.find_all("organization", recursive=False):
            organizations.append(element.get_text(strip=True))

        standard_descriptions = []
        for element in site.find_all("standard-description", recursive=False):
            standard_descriptions.append(element.get_text(strip=True))

        record = {
            "Condition_ID": condition_id,
            "Condition_Name": condition_name,
            "Resource_Title": resource_title,
            "Resource_URL": resource_url,
            "Information_Category": "; ".join(information_categories),
            "Organization": "; ".join(organizations),
            "Standard_Description": "; ".join(standard_descriptions),
            "Language_Mapped_URL": language_mapped_url
        }

        healthcare_resources.append(record)
healthcare_resources_df = pd.DataFrame(healthcare_resources)
display(healthcare_resources_df.head())
print("Total healthcare resources:", len(healthcare_resources_df))
print("Total columns:", len(healthcare_resources_df.columns))
```
